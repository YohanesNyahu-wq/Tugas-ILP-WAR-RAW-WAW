# Implementasi WAR dan Dependency Graph WAR

```python
# ==================================================
# TUGAS KELOMPOK - INSTRUCTION LEVEL PARALLELISM
# Kontributor : Yohanes
# Anggota     : 2
# Tugas       : Implementasi WAR dan Dependency Graph
# ==================================================

from dataclasses import dataclass
import matplotlib.pyplot as plt

@dataclass
class Instruction:
    name: str
    destination: str
    sources: list

# Contoh instruksi WAR
instructions = [
    Instruction("I1", "R4", ["R1", "R2"]),
    Instruction("I2", "R1", ["R3", "R5"])
]

def detect_war(inst1, inst2):
    return inst2.destination in inst1.sources

def main():
    print("IMPLEMENTASI WAR - YOHANES")

    for inst in instructions:
        print(f"{inst.name}: {inst.destination} = "
              f"{' + '.join(inst.sources)}")

    i1, i2 = instructions
    result = detect_war(i1, i2)

    print("\nHasil Analisis:")
    print("WAR terdeteksi:", result)

    assert result, "Pengujian WAR gagal"

    fig, ax = plt.subplots(figsize=(9, 3))

    ax.text(0.2, 0.5, "I1: R4 = R1 + R2",
            ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.5",
                      facecolor="lightblue"))

    ax.text(0.8, 0.5, "I2: R1 = R3 + R5",
            ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.5",
                      facecolor="lightyellow"))

    ax.annotate("", xy=(0.64, 0.5),
                xytext=(0.36, 0.5),
                arrowprops=dict(arrowstyle="->", lw=2))

    ax.text(0.5, 0.65, "WAR (R1)",
            ha="center", fontsize=11)

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")
    ax.set_title("Dependency Graph WAR - Yohanes")

    plt.savefig("war.png", dpi=200, bbox_inches="tight")
    plt.show()

if __name__ == "__main__":
    main()
```
