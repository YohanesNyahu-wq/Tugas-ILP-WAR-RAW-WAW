```python
# ==================================================
# TUGAS KELOMPOK - INSTRUCTION LEVEL PARALLELISM
# Kontributor : Irwansyah
# Anggota     : 3
# Tugas       : Implementasi WAW dan Dependency Graph
# ==================================================

from dataclasses import dataclass
import matplotlib.pyplot as plt

@dataclass
class Instruction:
    name: str
    destination: str
    sources: list

# Contoh instruksi WAW
instructions = [
    Instruction("I1", "R1", ["R2", "R3"]),
    Instruction("I2", "R1", ["R4", "R5"])
]

def detect_waw(inst1, inst2):
    return inst1.destination == inst2.destination

def main():
    print("IMPLEMENTASI WAW - IRWANSYAH")

    for inst in instructions:
        print(f"{inst.name}: {inst.destination} = "
              f"{' + '.join(inst.sources)}")

    i1, i2 = instructions
    result = detect_waw(i1, i2)

    print("\nHasil Analisis:")
    print("WAW terdeteksi:", result)

    assert result, "Pengujian WAW gagal"

    fig, ax = plt.subplots(figsize=(9, 3))

    ax.text(0.2, 0.5, "I1: R1 = R2 + R3",
            ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.5",
                      facecolor="lightblue"))

    ax.text(0.8, 0.5, "I2: R1 = R4 + R5",
            ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.5",
                      facecolor="lightyellow"))

    ax.annotate("", xy=(0.64, 0.5),
                xytext=(0.36, 0.5),
                arrowprops=dict(arrowstyle="->", lw=2))

    ax.text(0.5, 0.65, "WAW (R1)",
            ha="center", fontsize=11)

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")
    ax.set_title("Dependency Graph WAW - Irwansyah")

    plt.savefig("waw.png", dpi=200, bbox_inches="tight")
    plt.show()

if __name__ == "__main__":
    main()
```

```python
# ==================================================
# PENGUJIAN AKHIR RAW, WAR, WAW
# Kontributor : Irwansyah
# ==================================================

def detect_raw(a, b):
    return a.destination in b.sources

def detect_war(a, b):
    return b.destination in a.sources

def detect_waw(a, b):
    return a.destination == b.destination

test_cases = [
    ("RAW", detect_raw,
     Instruction("I1", "R1", ["R2", "R3"]),
     Instruction("I2", "R4", ["R1", "R5"])),

    ("WAR", detect_war,
     Instruction("I1", "R4", ["R1", "R2"]),
     Instruction("I2", "R1", ["R3", "R5"])),

    ("WAW", detect_waw,
     Instruction("I1", "R1", ["R2", "R3"]),
     Instruction("I2", "R1", ["R4", "R5"]))
]

print("HASIL PENGUJIAN KELOMPOK")

for jenis, fungsi, i1, i2 in test_cases:
    hasil = fungsi(i1, i2)
    assert hasil, f"Pengujian {jenis} gagal"
    print(f"{jenis}: LULUS")

print("\nSemua pengujian dependency berhasil.")
```
